```python
import os
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models, optimizers, losses
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.utils import shuffle
from imblearn.over_sampling import SMOTE
from scipy.io import wavfile
from scipy.signal import stft
import wfdb
from sklearn.metrics import classification_report, roc_auc_score, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd  # Added for PTB-XL CSV reading
import ast  # For parsing string lists in PTB-XL CSV
from tensorflow.keras import callbacks  # Added for early stopping if not already imported

# --- ECG Mamba Model Implementation (from ECGMamba paper) ---
# This is a minimal, self-contained Mamba block for 1D ECG data, inspired by the paper.
# If you have the official Mamba library, you can use it directly. Otherwise, this is a simplified version.

class SimpleMambaBlock(layers.Layer):
    def __init__(self, d_model, ssm_dim, dropout=0.5, **kwargs):
        super().__init__(**kwargs)
        self.d_model = d_model
        self.ssm_dim = ssm_dim
        self.dropout = dropout
        self.dense1 = layers.Dense(ssm_dim, activation='gelu', kernel_regularizer=tf.keras.regularizers.l2(1e-4))
        self.dense2 = layers.Dense(d_model, kernel_regularizer=tf.keras.regularizers.l2(1e-4))
        self.gate_dense = layers.Dense(d_model, activation='sigmoid', kernel_regularizer=tf.keras.regularizers.l2(1e-4))  # Gating mechanism      
        self.norm = layers.LayerNormalization(epsilon=1e-6)
        self.drop = layers.Dropout(dropout)

    def call(self, x, training=False):
        # SSM-like block: project, nonlinearity, project back, residual
        h = self.dense1(x)
        h = self.drop(h, training=training)
        h = self.dense2(h)
        h = self.drop(h, training=training)
        gate = self.gate_dense(x)  # Learn to select/weight features
        h = gate * h + (1 - gate) * x  # Selective residual connection
        return self.norm(h)

# ECG Mamba Model

def ecg_mamba_model(input_shape, d_model=128, ssm_dim=256, num_blocks=4, dropout=0.5):
    inputs = layers.Input(shape=input_shape)
    x = layers.Conv1D(d_model, kernel_size=3, padding='same', activation='gelu', kernel_regularizer=tf.keras.regularizers.l2(1e-4))(inputs)
    for _ in range(num_blocks):
        x = SimpleMambaBlock(d_model, ssm_dim, dropout)(x)
    x = layers.GlobalAveragePooling1D()(x)
    return models.Model(inputs, x, name="ecg_mamba_model")

# Multimodal Mamba Fusion Block
class MultimodalMambaBlock(layers.Layer):
    def __init__(self, d_model, ssm_dim, dropout=0.5, **kwargs):
        super().__init__(**kwargs)
        self.d_model = d_model
        self.ssm_dim = ssm_dim
        self.dropout = dropout
        self.dense1 = layers.Dense(ssm_dim, activation='gelu', kernel_regularizer=tf.keras.regularizers.l2(1e-4))
        self.dense2 = layers.Dense(d_model, kernel_regularizer=tf.keras.regularizers.l2(1e-4))
        self.gate_dense = layers.Dense(d_model, activation='sigmoid', kernel_regularizer=tf.keras.regularizers.l2(1e-4))  # Gating for fusion
        self.norm = layers.LayerNormalization(epsilon=1e-6)
        self.drop = layers.Dropout(dropout)

    def call(self, x, training=False):
        h = self.dense1(x)
        h = self.drop(h, training=training)
        h = self.dense2(h)
        h = self.drop(h, training=training)
        gate = self.gate_dense(x)
        h = gate * h + (1 - gate) * x
        return self.norm(h)

# Multimodal Mamba Fusion Model (now with selective fusion)
def multimodal_mamba_fusion(ecg_output, pcg_output, d_model=256, ssm_dim=512, num_blocks=2, dropout=0.5):
    # Paper-inspired selective fusion: learn a gate for each modality, then combine
    # 1. Project both modalities to same dim
    ecg_proj = layers.Dense(d_model, activation='gelu', kernel_regularizer=tf.keras.regularizers.l2(1e-4))(ecg_output)
    pcg_proj = layers.Dense(d_model, activation='gelu', kernel_regularizer=tf.keras.regularizers.l2(1e-4))(pcg_output)
    # 2. Learn a fusion gate (0-1) for each modality
    fusion_input = layers.Concatenate()([ecg_proj, pcg_proj])
    fusion_gate = layers.Dense(d_model, activation='sigmoid', kernel_regularizer=tf.keras.regularizers.l2(1e-4))(fusion_input)
    # 3. Weighted sum (selective fusion)
    fused = fusion_gate * ecg_proj + (1 - fusion_gate) * pcg_proj
    x = fused
    for _ in range(num_blocks):
        x = MultimodalMambaBlock(d_model, ssm_dim, dropout)(x)
    return x

# Constants
IMG_HEIGHT = 128
IMG_WIDTH = 128
PATCH_SIZE = 16
EMBED_DIM = 256
NUM_HEADS = 8
FF_DIM = 512
DROPOUT_RATE = 0.5
NUM_CLASSES = 2
INPUT_ECG_SHAPE = (500, 12)  # 12 leads instead of 1
INPUT_PCG_SHAPE = (IMG_HEIGHT, IMG_WIDTH, 1)
BATCH_SIZE = 32
EPOCHS = 65
LEARNING_RATE = 3e-4

# Helper function to compute PCG spectrogram
def compute_spectrogram(filepath):
    sample_rate, signal = wavfile.read(filepath)
    frequencies, times, spectrogram = stft(signal, fs=sample_rate, nperseg=256)
    spectrogram = np.log(np.abs(spectrogram) + 1e-10)
    spectrogram = np.resize(spectrogram, (IMG_HEIGHT, IMG_WIDTH))
    return spectrogram

# Load PCG dataset with augmentation for abnormal cases
def load_pcg_dataset(normal_dir, abnormal_dir):
    data, labels = [], []
    for label, directory in enumerate([normal_dir, abnormal_dir]):
        for filename in os.listdir(directory):
            if filename.endswith(".wav"):
                filepath = os.path.join(directory, filename)
                spectrogram = compute_spectrogram(filepath)
                data.append(spectrogram)
                labels.append(label)
                # Augment abnormal data by flipping
                if label == 1:
                    flipped_spectrogram = np.flip(spectrogram, axis=1)
                    data.append(flipped_spectrogram)
                    labels.append(label)
    return np.array(data), np.array(labels)

# --- PTB-XL ECG Data Loader (all leads) ---
def load_ptbxl_ecg_data(csv_path, records_path, num_samples=500):
    df = pd.read_csv(csv_path)
    available_records = []
    for root, dirs, files in os.walk(records_path):
        for file in files:
            if file.endswith('.hea'):
                base_id = file.split('_')[0]
                base_id = str(base_id).zfill(5)
                available_records.append(file.replace('.hea', ''))
    available_records = set(available_records)
    df = df[df['ecg_id'].astype(str).str.zfill(5).apply(lambda rid: any(rid in rec for rec in available_records))]
    X, Y = [], []
    exception_count = 0
    for idx, row in df.iterrows():
        rec_id = str(row['ecg_id']).zfill(5)
        rec_folder = rec_id[:2] + "000"
        rec_candidates = [rec for rec in available_records if rec.startswith(rec_id)]
        if not rec_candidates:
            if exception_count < 5:
                print(f"No matching record file for ecg_id {rec_id}")
            exception_count += 1
            continue
        rec_file = rec_candidates[0]
        rec_path = os.path.join(records_path, rec_folder, rec_file)
        try:
            signal, fields = wfdb.rdsamp(rec_path)
            # Use all 12 leads, shape: (time, 12)
            if signal.shape[0] < num_samples or signal.shape[1] < 12:
                continue
            # Take a random segment of num_samples
            start = np.random.randint(0, signal.shape[0] - num_samples + 1)
            segment = signal[start:start+num_samples, :12]  # shape (num_samples, 12)
            X.append(segment)
            scp_codes = ast.literal_eval(row['scp_codes']) if isinstance(row['scp_codes'], str) else row['scp_codes']
            label = 0 if 'NORM' in scp_codes else 1
            Y.append(label)
        except Exception as e:
            if exception_count < 5:
                print(f"Exception for record {rec_id}: {e}")
            exception_count += 1
            continue
    print(f"Loaded ECG samples: {len(X)}")
    print(f"Total exceptions: {exception_count}")
    return np.array(X), np.array(Y)

# --- Data Augmentation Utilities ---
import random

def add_ecg_noise(ecg, noise_level=0.05):
    noise = np.random.normal(0, noise_level, ecg.shape)
    return ecg + noise

def scale_ecg(ecg, scale_range=(0.9, 1.1)):
    scale = np.random.uniform(*scale_range)
    return ecg * scale

def time_mask_ecg(ecg, max_mask=20):
    ecg = ecg.copy()
    mask_len = np.random.randint(1, max_mask)
    start = np.random.randint(0, ecg.shape[0] - mask_len)
    ecg[start:start+mask_len, :] = 0
    return ecg

# --- Aggressive ECG Augmentation: time warping, random erasing, cutmix ---
def augment_ecg(ecg):
    # Add noise
    if np.random.rand() < 0.5:
        ecg = add_ecg_noise(ecg)
    # Scale
    if np.random.rand() < 0.5:
        ecg = scale_ecg(ecg)
    # Time masking
    if np.random.rand() < 0.3:
        ecg = time_mask_ecg(ecg)
    # Time warping (random stretch/compress)
    if np.random.rand() < 0.3:
        orig_len = ecg.shape[0]
        warp_factor = np.random.uniform(0.8, 1.2)
        new_len = int(orig_len * warp_factor)
        ecg = np.array([np.interp(np.linspace(0, orig_len-1, new_len), np.arange(orig_len), ecg[:,i]) for i in range(ecg.shape[1])]).T
        if new_len > orig_len:
            ecg = ecg[:orig_len, :]
        else:
            pad = np.zeros((orig_len-new_len, ecg.shape[1]))
            ecg = np.concatenate([ecg, pad], axis=0)
    # Random erasing (zero out a random segment)
    if np.random.rand() < 0.2:
        seg_len = np.random.randint(10, 40)
        start = np.random.randint(0, ecg.shape[0] - seg_len)
        ecg[start:start+seg_len, :] = 0
    return ecg

def freq_mask_spectrogram(spec, max_mask=20):
    spec = spec.copy()
    num_freqs = spec.shape[0]
    mask_len = np.random.randint(1, max_mask)
    start = np.random.randint(0, num_freqs - mask_len)
    spec[start:start+mask_len, :] = 0
    return spec

def time_mask_spectrogram(spec, max_mask=20):
    spec = spec.copy()
    num_times = spec.shape[1]
    mask_len = np.random.randint(1, max_mask)
    start = np.random.randint(0, num_times - mask_len)
    spec[:, start:start+mask_len] = 0
    return spec

# --- Aggressive Spectrogram Augmentation: time/freq warping, random erasing, cutmix ---
def augment_spectrogram(spec):
    # Frequency masking
    if np.random.rand() < 0.5:
        spec = freq_mask_spectrogram(spec)
    # Time masking
    if np.random.rand() < 0.5:
        spec = time_mask_spectrogram(spec)
    # Time warping (stretch/compress along time axis)
    if np.random.rand() < 0.3:
        orig_len = spec.shape[1]
        warp_factor = np.random.uniform(0.8, 1.2)
        new_len = int(orig_len * warp_factor)
        spec = np.array([np.interp(np.linspace(0, orig_len-1, new_len), np.arange(orig_len), spec[i,:]) for i in range(spec.shape[0])])
        if new_len > orig_len:
            spec = spec[:, :orig_len]
        else:
            pad = np.zeros((spec.shape[0], orig_len-new_len))
            spec = np.concatenate([spec, pad], axis=1)
    # Random erasing (zero out a random patch)
    if np.random.rand() < 0.2:
        freq_len = np.random.randint(5, 20)
        time_len = np.random.randint(5, 20)
        f0 = np.random.randint(0, spec.shape[0] - freq_len)
        t0 = np.random.randint(0, spec.shape[1] - time_len)
        spec[f0:f0+freq_len, t0:t0+time_len] = 0
    return spec

# --- CutMix for ECG and PCG ---
def cutmix_ecg(ecg1, ecg2):
    # Randomly select a segment and replace with another sample
    seg_len = np.random.randint(20, 80)
    start = np.random.randint(0, ecg1.shape[0] - seg_len)
    ecg = ecg1.copy()
    ecg[start:start+seg_len, :] = ecg2[start:start+seg_len, :]
    return ecg

def cutmix_spectrogram(spec1, spec2):
    # Randomly select a patch and replace with another sample
    freq_len = np.random.randint(10, 40)
    time_len = np.random.randint(10, 40)
    f0 = np.random.randint(0, spec1.shape[0] - freq_len)
    t0 = np.random.randint(0, spec1.shape[1] - time_len)
    spec = spec1.copy()
    spec[f0:f0+freq_len, t0:t0+time_len] = spec2[f0:f0+freq_len, t0:t0+time_len]
    return spec

# Mixup for ECG and PCG
def mixup(x1, y1, x2, y2, alpha=0.2):
    lam = np.random.beta(alpha, alpha)
    x = lam * x1 + (1 - lam) * x2
    y = lam * y1 + (1 - lam) * y2
    return x, y

# Data paths
pcg_normal_dir = "/kaggle/input/datasets/mahirazmainhaque/combined-heart-sound-dataphysionet-circor/combined_heart_sound_data(physionet +circor)/Normal"
pcg_abnormal_dir = "/kaggle/input/datasets/mahirazmainhaque/combined-heart-sound-dataphysionet-circor/combined_heart_sound_data(physionet +circor)/Abnormal"
# Use PTB-XL dataset path for ECG
ptbxl_data_path = "/kaggle/input/datasets/khyeh0719/ptb-xl-dataset/ptb-xl-a-large-publicly-available-electrocardiography-dataset-1.0.1"
ptbxl_csv_path = os.path.join(ptbxl_data_path, "ptbxl_database.csv")
ptbxl_records_path = os.path.join(ptbxl_data_path, "records100")


early_stop = callbacks.EarlyStopping(monitor='val_loss', patience=7, restore_best_weights=True)

# Load and preprocess PCG data
pcg_data, pcg_labels = load_pcg_dataset(pcg_normal_dir, pcg_abnormal_dir)
pcg_data = pcg_data / np.max(pcg_data)
pcg_data = np.expand_dims(pcg_data, -1)
pcg_labels = tf.keras.utils.to_categorical(pcg_labels, NUM_CLASSES)

# Load and preprocess ECG data using PTB-XL (all leads)
X_clean, Y_clean = load_ptbxl_ecg_data(ptbxl_csv_path, ptbxl_records_path, num_samples=500)
scaler = StandardScaler()
X_clean = np.array([scaler.fit_transform(x) for x in X_clean])  # shape (n, 500, 12)
Y_clean = np.array(Y_clean)

# Apply SMOTE to balance ECG dataset
X_clean_reshaped = X_clean.reshape((X_clean.shape[0], -1))  # shape (n, 2400)
smote = SMOTE(random_state=42)
X_clean_bal, Y_clean_bal = smote.fit_resample(X_clean_reshaped, Y_clean)
Y_clean_bal = tf.keras.utils.to_categorical(Y_clean_bal, NUM_CLASSES)
X_clean_bal = X_clean_bal.reshape((-1, 500, 12))

# --- Apply Data Augmentation to Datasets ---
# Augment PCG data (spectrograms) with more aggressive augmentations
aug_pcg_data = []
aug_pcg_labels = []
for i in range(len(pcg_data)):
    spec = pcg_data[i][..., 0]  # Remove channel for augmentation
    label = pcg_labels[i]
    aug_pcg_data.append(pcg_data[i])  # original
    aug_pcg_labels.append(label)
    # Standard augmentation
    if np.random.rand() < 0.5:
        aug_spec = augment_spectrogram(spec)
        aug_pcg_data.append(np.expand_dims(aug_spec, -1))
        aug_pcg_labels.append(label)
    # CutMix augmentation
    if np.random.rand() < 0.2:
        j = np.random.randint(0, len(pcg_data))
        spec2 = pcg_data[j][..., 0]
        cutmix_spec = cutmix_spectrogram(spec, spec2)
        aug_pcg_data.append(np.expand_dims(cutmix_spec, -1))
        aug_pcg_labels.append(label)

pcg_data = np.array(aug_pcg_data)
pcg_labels = np.array(aug_pcg_labels)

# Augment ECG data (all leads) with more aggressive augmentations
aug_ecg_data = []
aug_ecg_labels = []
for i in range(len(X_clean_bal)):
    ecg = X_clean_bal[i]
    label = Y_clean_bal[i]
    aug_ecg_data.append(ecg)  # original
    aug_ecg_labels.append(label)
    # Standard augmentation
    if np.random.rand() < 0.5:
        aug_ecg = augment_ecg(ecg)
        aug_ecg_data.append(aug_ecg)
        aug_ecg_labels.append(label)
    # CutMix augmentation
    if np.random.rand() < 0.2:
        j = np.random.randint(0, len(X_clean_bal))
        ecg2 = X_clean_bal[j]
        cutmix_ecg_aug = cutmix_ecg(ecg, ecg2)
        aug_ecg_data.append(cutmix_ecg_aug)
        aug_ecg_labels.append(label)

X_clean_bal = np.array(aug_ecg_data)
Y_clean_bal = np.array(aug_ecg_labels)

# Balance datasets
min_samples = min(len(X_clean_bal), len(pcg_data))
X_clean_bal, Y_clean_bal = shuffle(X_clean_bal[:min_samples], Y_clean_bal[:min_samples], random_state=42)
pcg_data, pcg_labels = shuffle(pcg_data[:min_samples], pcg_labels[:min_samples], random_state=42)

# Train-test split
x_train_pcg, x_test_pcg, y_train_pcg, y_test_pcg = train_test_split(pcg_data, pcg_labels, test_size=0.1, random_state=42)
x_val_pcg, x_test_pcg, y_val_pcg, y_test_pcg = train_test_split(x_test_pcg, y_test_pcg, test_size=0.2, random_state=42)

x_train_ecg, x_test_ecg, y_train_ecg, y_test_ecg = train_test_split(X_clean_bal, Y_clean_bal, test_size=0.1, random_state=42)
x_val_ecg, x_test_ecg, y_val_ecg, y_test_ecg = train_test_split(x_test_ecg, y_test_ecg, test_size=0.2, random_state=42)

# Patch Embedding Layer for PCG
def patch_embedding(input_shape, patch_size, embed_dim):
    inputs = layers.Input(shape=input_shape)
    patches = layers.Conv2D(embed_dim, (patch_size, patch_size), strides=(patch_size, patch_size), kernel_regularizer=tf.keras.regularizers.l2(1e-4))(inputs)
    flat_patches = layers.Reshape((-1, embed_dim))(patches)
    return models.Model(inputs, flat_patches, name="patch_embedding")

# Transformer Block
def transformer_block(embed_dim, num_heads, ff_dim, dropout_rate):
    inputs = layers.Input(shape=(None, embed_dim))
    attention_output = layers.MultiHeadAttention(num_heads=num_heads, key_dim=embed_dim)(inputs, inputs)
    attention_output = layers.Dropout(dropout_rate)(attention_output)
    attention_output = layers.LayerNormalization(epsilon=1e-6)(inputs + attention_output)

    ffn_output = layers.Dense(ff_dim, activation='relu', kernel_regularizer=tf.keras.regularizers.l2(1e-4))(attention_output)
    ffn_output = layers.Dense(embed_dim, kernel_regularizer=tf.keras.regularizers.l2(1e-4))(ffn_output)
    ffn_output = layers.Dropout(dropout_rate)(ffn_output)
    outputs = layers.LayerNormalization(epsilon=1e-6)(attention_output + ffn_output)

    return models.Model(inputs, outputs, name="transformer_block")

# PCG Transformer Model
def pcg_transformer(input_shape, patch_size, embed_dim, num_heads, ff_dim, dropout_rate):
    inputs = layers.Input(shape=input_shape)
    x = patch_embedding(input_shape, patch_size, embed_dim)(inputs)
    x = transformer_block(embed_dim, num_heads, ff_dim, dropout_rate)(x)
    x = layers.GlobalAveragePooling1D()(x)
    return models.Model(inputs, x, name="pcg_transformer")

# Final Hybrid Model
def hybrid_transformer_model():
    ecg_input = layers.Input(shape=INPUT_ECG_SHAPE, name="ECG_Input")
    pcg_input = layers.Input(shape=INPUT_PCG_SHAPE, name="PCG_Input")

    # Use ECG Mamba model instead of transformer
    ecg_model = ecg_mamba_model(INPUT_ECG_SHAPE, d_model=256, ssm_dim=512, num_blocks=8, dropout=0.5)
    pcg_model = pcg_transformer(INPUT_PCG_SHAPE, PATCH_SIZE, EMBED_DIM, NUM_HEADS, FF_DIM, 0.5)

    ecg_output = ecg_model(ecg_input)
    pcg_output = pcg_model(pcg_input)

    # Use Multimodal Mamba Fusion only
    fusion_output = multimodal_mamba_fusion(ecg_output, pcg_output, d_model=EMBED_DIM, ssm_dim=512, num_blocks=2, dropout=0.5)

    # Remove Stacked Transformer Ensemble
    # Combine Fusion Output directly
    outputs = layers.Dense(NUM_CLASSES, activation='sigmoid', kernel_regularizer=tf.keras.regularizers.l2(1e-4))(fusion_output)

    model = models.Model(inputs=[ecg_input, pcg_input], outputs=outputs, name="Hybrid_Transformer")
    model.compile(optimizer=optimizers.Adam(learning_rate=LEARNING_RATE),
                  loss=losses.BinaryCrossentropy(),
                  metrics=['accuracy'])

    return model

# --- Mixup Regularization for Training Batches ---
def mixup_batch(x1, x2, y, alpha=0.2):
    batch_size = x1.shape[0]
    indices = np.random.permutation(batch_size)
    x1_shuffled = x1[indices]
    x2_shuffled = x2[indices]
    y_shuffled = y[indices]
    lam = np.random.beta(alpha, alpha, batch_size)
    # For ECG (3D): (batch, 1, 1), for PCG (4D): (batch, 1, 1, 1)
    lam_ecg = lam.reshape(-1, 1, 1)
    lam_pcg = lam.reshape(-1, 1, 1, 1)
    x1_mix = lam_ecg * x1 + (1 - lam_ecg) * x1_shuffled
    x2_mix = lam_pcg * x2 + (1 - lam_pcg) * x2_shuffled
    y_mix = lam.reshape(-1, 1) * y + (1 - lam).reshape(-1, 1) * y_shuffled
    return x1_mix, x2_mix, y_mix

# Custom data generator with Mixup
def mixup_generator(x1, x2, y, batch_size=64, alpha=0.2):
    num_samples = x1.shape[0]
    indices = np.arange(num_samples)
    while True:
        np.random.shuffle(indices)
        for i in range(0, num_samples, batch_size):
            batch_idx = indices[i:i+batch_size]
            if len(batch_idx) < batch_size:
                continue  # Skip last batch if it's smaller than batch_size
            x1_batch = x1[batch_idx]
            x2_batch = x2[batch_idx]
            y_batch = y[batch_idx]
            x1_mix, x2_mix, y_mix = mixup_batch(x1_batch, x2_batch, y_batch, alpha)
            yield (x1_mix, x2_mix), y_mix

# --- Use Mixup Generator for Training ---
train_gen = tf.data.Dataset.from_generator(
    lambda: mixup_generator(x_train_ecg, x_train_pcg, y_train_ecg, batch_size=BATCH_SIZE, alpha=0.2),
    output_signature=(
        (
            tf.TensorSpec(shape=(BATCH_SIZE, 500, 12), dtype=tf.float32),
            tf.TensorSpec(shape=(BATCH_SIZE, 128, 128, 1), dtype=tf.float32)
        ),
        tf.TensorSpec(shape=(BATCH_SIZE, NUM_CLASSES), dtype=tf.float32)
    )
)
val_gen = tf.data.Dataset.from_generator(
    lambda: mixup_generator(x_val_ecg, x_val_pcg, y_val_ecg, batch_size=BATCH_SIZE, alpha=0.2),
    output_signature=(
        (
            tf.TensorSpec(shape=(BATCH_SIZE, 500, 12), dtype=tf.float32),
            tf.TensorSpec(shape=(BATCH_SIZE, 128, 128, 1), dtype=tf.float32)
        ),
        tf.TensorSpec(shape=(BATCH_SIZE, NUM_CLASSES), dtype=tf.float32)
    )
)
steps_per_epoch = len(x_train_ecg) // BATCH_SIZE
validation_steps = len(x_val_ecg) // BATCH_SIZE

model = hybrid_transformer_model()
history = model.fit(
    train_gen,
    steps_per_epoch=steps_per_epoch,
    validation_data=val_gen,
    validation_steps=validation_steps,
    epochs=EPOCHS,
    callbacks=[early_stop]  # Added early stopping
)
y_pred_prob = model.predict([x_test_ecg, x_test_pcg])
y_pred = (y_pred_prob[:, 1] > 0.57).astype(int)  # Adjusted threshold for better recall
y_true = np.argmax(y_test_ecg, axis=1)

print("Classification Report:")
print(classification_report(y_true, y_pred))
print("AUC-ROC:", roc_auc_score(y_true, y_pred_prob[:, 1]))

plt.figure(figsize=(5, 5))
sns.heatmap(confusion_matrix(y_true, y_pred), annot=True, fmt='d', cmap='Blues', xticklabels=['Normal', 'Abnormal'], yticklabels=['Normal', 'Abnormal'])
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.show()

model.summary()
```

```python
# --- Federated Learning with Weighted Averaging, ROC-AUC Tracking, and Checkpointing ---
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
from sklearn.metrics import roc_auc_score
from sklearn.utils.class_weight import compute_class_weight

# --- Helper: safely flatten labels regardless of shape (one-hot, (N,1), or (N,)) ---
def get_flat_labels(y):
    """Convert y to a flat 1D array of integer class labels, regardless of input shape."""
    y = np.asarray(y)
    if y.ndim > 1 and y.shape[-1] > 1:
        # one-hot / softmax-style labels -> convert to class indices
        return np.argmax(y, axis=-1)
    else:
        # shape (N,1) or (N,) -> just squeeze to 1D
        return y.reshape(-1).astype(int)

# --- Partition data for federated simulation ---
def partition_data(x_ecg, x_pcg, y, num_clients=3):
    size = len(y) // num_clients
    partitions = []
    for i in range(num_clients):
        start = i * size
        end = (i + 1) * size if i < num_clients - 1 else len(y)
        partitions.append((x_ecg[start:end], x_pcg[start:end], y[start:end]))
    return partitions

num_clients = 3
client_partitions = partition_data(x_train_ecg, x_train_pcg, y_train_ecg, num_clients=num_clients)

# --- Weighted FedAvg: weight each client's update by its sample count ---
def weighted_average_weights(weight_list, sample_counts):
    total = sum(sample_counts)
    avg_weights = []
    for layer_weights in zip(*weight_list):
        stacked = np.array([w * (n / total) for w, n in zip(layer_weights, sample_counts)])
        avg_weights.append(np.sum(stacked, axis=0))
    return avg_weights

# --- Per-round LR schedule (mild decay each round) ---
def get_round_lr(base_lr, rnd, decay=0.9):
    return base_lr * (decay ** rnd)

num_rounds = 7          # more rounds = more chances to converge toward higher AUC
local_epochs = 3
batch_size = 16
base_lr = 1e-3
patience = 5              # early stop if AUC doesn't improve
min_delta = 1e-4

global_model = hybrid_transformer_model()
global_weights = global_model.get_weights()

round_accuracies = []
round_losses = []
round_aucs = []

best_auc = -np.inf
best_weights = global_weights
rounds_no_improve = 0

# Pre-flatten test labels once for AUC computation
y_test_flat = get_flat_labels(y_test_ecg)
n_test_classes = len(np.unique(y_test_flat))

for rnd in range(num_rounds):
    print(f"Federated Round {rnd+1}/{num_rounds}")
    client_weights = []
    sample_counts = []
    current_lr = get_round_lr(base_lr, rnd)

    for cid in range(num_clients):
        x_ecg, x_pcg, y = client_partitions[cid]

        # Handle local class imbalance per client (flatten labels just for weight calc)
        y_flat = get_flat_labels(y)
        classes = np.unique(y_flat)

        if len(classes) > 1:
            cw = compute_class_weight('balanced', classes=classes, y=y_flat)
            class_weight = {int(c): w for c, w in zip(classes, cw)}
        else:
            class_weight = None

        model = hybrid_transformer_model()
        model.set_weights(global_weights)
        model.optimizer.learning_rate.assign(current_lr)

        model.fit(
            [x_ecg, x_pcg], y,           # original y (one-hot or whatever shape the model expects)
            epochs=local_epochs,
            batch_size=batch_size,
            class_weight=class_weight,
            verbose=0
        )
        client_weights.append(model.get_weights())
        sample_counts.append(len(y))

    global_weights = weighted_average_weights(client_weights, sample_counts)
    global_model.set_weights(global_weights)

    # --- Evaluate on test set after each round ---
    loss, acc = global_model.evaluate([x_test_ecg, x_test_pcg], y_test_ecg, verbose=0)

    # --- ROC-AUC: handle binary vs multi-class output shapes safely ---
    y_pred_probs = global_model.predict([x_test_ecg, x_test_pcg], verbose=0)

    if y_pred_probs.shape[-1] == 1:
        # Binary sigmoid output
        auc = roc_auc_score(y_test_flat, y_pred_probs.ravel())
    elif n_test_classes == 2 and y_pred_probs.shape[-1] == 2:
        # Softmax binary case -> use probability of positive class
        auc = roc_auc_score(y_test_flat, y_pred_probs[:, 1])
    else:
        # True multi-class -> one-vs-rest AUC, needs integer labels + full prob matrix
        auc = roc_auc_score(y_test_flat, y_pred_probs, multi_class='ovr')

    round_accuracies.append(acc)
    round_losses.append(loss)
    round_aucs.append(auc)

    print(f"  LR: {current_lr:.6f} | Test Loss: {loss:.4f} | Test Acc: {acc:.4f} | Test ROC-AUC: {auc:.4f}")

    # --- Checkpoint the best round by AUC, not just the last round ---
    if auc - best_auc > min_delta:
        best_auc = auc
        best_weights = global_weights
        rounds_no_improve = 0
    else:
        rounds_no_improve += 1

    if rounds_no_improve >= patience:
        print(f"No AUC improvement for {patience} rounds. Stopping early at round {rnd+1}.")
        break

# --- Restore best-performing weights before saving ---
global_model.set_weights(best_weights)
global_model.save('Multimodal_Mamba_Federated_Model.keras')
print(f"Federated model saved. Best Test ROC-AUC achieved: {best_auc:.4f}")

# --- Plot accuracy, loss, and ROC-AUC curves ---
completed_rounds = len(round_accuracies)
plt.figure(figsize=(15, 4))

plt.subplot(1, 3, 1)
plt.plot(range(1, completed_rounds + 1), round_accuracies, marker='o')
plt.title('Test Accuracy per Round')
plt.xlabel('Federated Round')
plt.ylabel('Accuracy')
plt.grid(True)

plt.subplot(1, 3, 2)
plt.plot(range(1, completed_rounds + 1), round_losses, marker='o', color='orange')
plt.title('Test Loss per Round')
plt.xlabel('Federated Round')
plt.ylabel('Loss')
plt.grid(True)

plt.subplot(1, 3, 3)
plt.plot(range(1, completed_rounds + 1), round_aucs, marker='o', color='green')
plt.axhline(y=best_auc, color='red', linestyle='--', label=f'Best AUC: {best_auc:.4f}')
plt.title('Test ROC-AUC per Round')
plt.xlabel('Federated Round')
plt.ylabel('ROC-AUC')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# --- Independent Mamba+Transformer Branches (ECG, PCG) + Federated Learning + Decision Fusion ---
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, Model
import matplotlib.pyplot as plt
from sklearn.metrics import roc_auc_score
from sklearn.utils.class_weight import compute_class_weight
from tensorflow.keras.utils import to_categorical

# ============================================================
# 1. MAMBA (SELECTIVE SSM) BLOCK — tf.scan based
# ============================================================
class MambaBlock(layers.Layer):
    def __init__(self, d_model, d_state=16, expand=2, **kwargs):
        super().__init__(**kwargs)
        self.d_model = d_model
        self.d_state = d_state
        self.d_inner = expand * d_model

        self.in_proj = layers.Dense(self.d_inner * 2, use_bias=False)
        self.conv1d = layers.Conv1D(
            filters=self.d_inner, kernel_size=4, padding='causal',
            groups=self.d_inner, use_bias=True, activation='swish'
        )
        self.x_proj = layers.Dense(self.d_state * 2 + self.d_inner, use_bias=False)
        self.dt_proj = layers.Dense(self.d_inner, use_bias=True)
        self.out_proj = layers.Dense(d_model, use_bias=False)

        self.A_log = self.add_weight(
            shape=(self.d_inner, self.d_state), initializer='ones',
            trainable=True, name='A_log'
        )
        self.D = self.add_weight(
            shape=(self.d_inner,), initializer='ones', trainable=True, name='D'
        )

    def call(self, x):
        batch = tf.shape(x)[0]

        x_and_res = self.in_proj(x)
        x_in, res = tf.split(x_and_res, 2, axis=-1)
        x_in = self.conv1d(x_in)

        x_dbc = self.x_proj(x_in)
        delta_raw, B, C = tf.split(
            x_dbc, [self.d_inner, self.d_state, self.d_state], axis=-1
        )
        delta = tf.nn.softplus(self.dt_proj(delta_raw))

        A = -tf.exp(self.A_log)

        deltaA = tf.exp(tf.einsum('bld,dn->bldn', delta, A))
        deltaB_x = tf.einsum('bld,bln,bld->bldn', delta, B, x_in)

        deltaA_t = tf.transpose(deltaA, [1, 0, 2, 3])
        deltaB_x_t = tf.transpose(deltaB_x, [1, 0, 2, 3])

        def scan_fn(h_prev, elems):
            dA_t, dBx_t = elems
            return dA_t * h_prev + dBx_t

        h0 = tf.zeros((batch, self.d_inner, self.d_state))
        h_states = tf.scan(scan_fn, (deltaA_t, deltaB_x_t), initializer=h0)
        h_states = tf.transpose(h_states, [1, 0, 2, 3])

        y = tf.einsum('bldn,bln->bld', h_states, C) + self.D * x_in
        y = y * tf.nn.swish(res)

        return self.out_proj(y)


# ============================================================
# 2. TRANSFORMER ENCODER BLOCK
# ============================================================
def transformer_encoder_block(x, d_model, num_heads=4, ff_dim=128, dropout=0.1):
    attn_out = layers.MultiHeadAttention(num_heads=num_heads, key_dim=d_model // num_heads)(x, x)
    attn_out = layers.Dropout(dropout)(attn_out)
    x = layers.LayerNormalization(epsilon=1e-6)(x + attn_out)

    ff = layers.Dense(ff_dim, activation='relu')(x)
    ff = layers.Dense(d_model)(ff)
    ff = layers.Dropout(dropout)(ff)
    return layers.LayerNormalization(epsilon=1e-6)(x + ff)


# ============================================================
# 3. UNIMODAL HYBRID MAMBA+TRANSFORMER MODEL BUILDERS
#    (each is a STANDALONE model: single input, single output —
#     no cross-modality concatenation, no assumption of pairing)
# ============================================================
def build_unimodal_model(input_shape, d_model=64, num_classes=2, lr=1e-3, name_prefix="branch"):
    inp = layers.Input(shape=input_shape, name=f"{name_prefix}_input")

    if len(input_shape) == 3:
        h, w, c = input_shape
        x = layers.Reshape((h, w * c))(inp)
    else:
        x = inp

    x = layers.Dense(d_model)(x)
    x = MambaBlock(d_model=d_model, d_state=16)(x)
    x = transformer_encoder_block(x, d_model)
    x = layers.GlobalAveragePooling1D()(x)
    x = layers.Dense(64, activation='relu')(x)
    x = layers.Dropout(0.3)(x)
    output = layers.Dense(num_classes, activation='softmax')(x)

    model = Model(inputs=inp, outputs=output)
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=lr),
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )
    return model

def build_ecg_model(input_shape=(500, 12), d_model=64):
    return build_unimodal_model(input_shape, d_model=d_model, name_prefix="ecg")

def build_pcg_model(input_shape=(128, 128, 1), d_model=64):
    return build_unimodal_model(input_shape, d_model=d_model, name_prefix="pcg")


# ============================================================
# 4. GENERIC HELPERS
# ============================================================
def get_flat_labels(y):
    y = np.asarray(y)
    if y.ndim > 1 and y.shape[-1] > 1:
        return np.argmax(y, axis=-1)
    return y.reshape(-1).astype(int)

def non_iid_partition(x, y, num_clients=2, classes_per_client=1):
    y = np.array(y)
    unique_labels = np.unique(y)
    partitions = [[] for _ in range(num_clients)]
    label_indices = {label: np.where(y == label)[0] for label in unique_labels}
    for idxs in label_indices.values():
        np.random.shuffle(idxs)
    label_splits = np.array_split(unique_labels, num_clients)
    for client_id, client_labels in enumerate(label_splits):
        client_idxs = []
        for label in client_labels:
            n = len(label_indices[label]) // classes_per_client
            client_idxs.extend(label_indices[label][:n])
            label_indices[label] = label_indices[label][n:]
        if len(client_idxs) == 0:
            partitions[client_id] = (np.array([]), np.array([]))
            continue
        client_idxs = np.array(client_idxs, dtype=int)
        partitions[client_id] = (x[client_idxs], y[client_idxs])
    return partitions

def weighted_average_weights(weight_list, sample_counts):
    total = sum(sample_counts)
    avg_weights = []
    for layer_weights in zip(*weight_list):
        stacked = np.array([w * (n / total) for w, n in zip(layer_weights, sample_counts)])
        avg_weights.append(np.sum(stacked, axis=0))
    return avg_weights

def get_round_lr(base_lr, rnd, decay=0.9):
    return base_lr * (decay ** rnd)


# ============================================================
# 5. GENERIC FEDERATED TRAINING LOOP FOR ONE MODALITY
# ============================================================
def run_federated_branch(build_model_fn, x_train, y_train, x_test, y_test,
                          modality_name, num_clients=2, classes_per_client=1,
                          num_rounds=15, local_epochs=3, batch_size=16,
                          base_lr=1e-3, patience=5, min_delta=1e-4):

    y_train_1d = get_flat_labels(y_train)
    client_partitions = non_iid_partition(x_train, y_train_1d, num_clients, classes_per_client)

    for i in range(len(client_partitions)):
        xc, yc = client_partitions[i]
        if len(yc) > 0:
            client_partitions[i] = (xc, to_categorical(yc, num_classes=2))

    y_test_oh = to_categorical(get_flat_labels(y_test), num_classes=2)
    y_test_flat = get_flat_labels(y_test_oh)

    global_model = build_model_fn()
    global_weights = global_model.get_weights()

    best_auc, best_weights, rounds_no_improve = -np.inf, global_weights, 0
    history = {"acc": [], "loss": [], "auc": []}

    for rnd in range(num_rounds):
        print(f"[{modality_name}] Round {rnd+1}/{num_rounds}")
        client_weights, sample_counts = [], []
        current_lr = get_round_lr(base_lr, rnd)

        for cid in range(num_clients):
            xc, yc = client_partitions[cid]
            if len(xc) == 0:
                continue
            yc_flat = get_flat_labels(yc)
            classes = np.unique(yc_flat)
            class_weight = None
            if len(classes) > 1:
                cw = compute_class_weight('balanced', classes=classes, y=yc_flat)
                class_weight = {int(c): w for c, w in zip(classes, cw)}

            model = build_model_fn()
            model.set_weights(global_weights)
            model.optimizer.learning_rate.assign(current_lr)
            model.fit(xc, yc, epochs=local_epochs, batch_size=batch_size,
                      class_weight=class_weight, verbose=0)
            client_weights.append(model.get_weights())
            sample_counts.append(len(yc))

        if not client_weights:
            print("  No clients participated.")
            continue

        global_weights = weighted_average_weights(client_weights, sample_counts)
        global_model.set_weights(global_weights)

        loss, acc = global_model.evaluate(x_test, y_test_oh, verbose=0)
        probs = global_model.predict(x_test, verbose=0)
        auc = roc_auc_score(y_test_flat, probs[:, 1]) if probs.shape[-1] == 2 else roc_auc_score(y_test_flat, probs.ravel())

        history["acc"].append(acc); history["loss"].append(loss); history["auc"].append(auc)
        print(f"  LR:{current_lr:.6f} Loss:{loss:.4f} Acc:{acc:.4f} AUC:{auc:.4f}")

        if auc - best_auc > min_delta:
            best_auc, best_weights, rounds_no_improve = auc, global_weights, 0
        else:
            rounds_no_improve += 1
        if rounds_no_improve >= patience:
            print(f"  Early stop at round {rnd+1}")
            break

    global_model.set_weights(best_weights)
    return global_model, history, best_auc, y_test_flat, probs


# ============================================================
# 6. RUN BOTH BRANCHES INDEPENDENTLY
#    (x_train_ecg/y_train_ecg and x_train_pcg/y_train_pcg must come
#     from SEPARATE train_test_splits — do NOT truncate to min_samples)
# ============================================================
ecg_model, ecg_history, ecg_best_auc, ecg_y_flat, ecg_probs = run_federated_branch(
    build_ecg_model, x_train_ecg, y_train_ecg, x_test_ecg, y_test_ecg,
    modality_name="ECG", num_clients=3, classes_per_client=1
)

pcg_model, pcg_history, pcg_best_auc, pcg_y_flat, pcg_probs = run_federated_branch(
    build_pcg_model, x_train_pcg, y_train_pcg, x_test_pcg, y_test_pcg,
    modality_name="PCG", num_clients=3, classes_per_client=1
)

print(f"\nFinal ECG-branch Test ROC-AUC: {ecg_best_auc:.4f}")
print(f"Final PCG-branch Test ROC-AUC: {pcg_best_auc:.4f}")

ecg_model.save('ECG_Mamba_Federated_Model.keras')
pcg_model.save('PCG_Mamba_Federated_Model.keras')

# ============================================================
# 7. DECISION-LEVEL FUSION — ONLY VALID WITH GENUINELY PAIRED TEST PATIENTS
# ============================================================
# Do not compute a "fused AUC" on x_test_ecg/x_test_pcg together — they are
# still unrelated patients even at test time under the current data loaders.
#
# If you obtain (or already have, and haven't used) a small set of patients
# with BOTH an ECG and a PCG recording, evaluate fusion only on that subset:
#
# ecg_probs_paired = ecg_model.predict(x_paired_ecg, verbose=0)[:, 1]
# pcg_probs_paired = pcg_model.predict(x_paired_pcg, verbose=0)[:, 1]
# fused_probs = 0.5 * ecg_probs_paired + 0.5 * pcg_probs_paired
# fused_auc = roc_auc_score(y_paired, fused_probs)
# print(f"Fused (paired-patient) ROC-AUC: {fused_auc:.4f}")

plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(ecg_history["auc"], marker='o', label='ECG (Mamba+Transformer)')
plt.plot(pcg_history["auc"], marker='s', label='PCG (Mamba+Transformer)')
plt.title('ROC-AUC per Round (Independent Branches)')
plt.xlabel('Round'); plt.ylabel('ROC-AUC'); plt.legend(); plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(ecg_history["loss"], marker='o', label='ECG')
plt.plot(pcg_history["loss"], marker='s', label='PCG')
plt.title('Loss per Round (Independent Branches)')
plt.xlabel('Round'); plt.ylabel('Loss'); plt.legend(); plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score
import numpy as np

def evaluate_branch(model, x_test, y_test_onehot, branch_name):
    """
    Full evaluation for one unimodal branch: classification report,
    confusion matrix, and ROC-AUC.
    """
    y_pred_prob = model.predict(x_test, verbose=0)
    y_true = y_test_onehot.argmax(axis=1)
    y_pred = y_pred_prob.argmax(axis=1)

    print(f"\n{'='*50}")
    print(f"{branch_name} Branch Evaluation")
    print(f"{'='*50}")

    print("\nClassification Report:")
    print(classification_report(y_true, y_pred, target_names=["Normal", "Abnormal"]))

    print("Confusion Matrix:")
    cm = confusion_matrix(y_true, y_pred)
    print(cm)

    if y_pred_prob.shape[1] == 2:
        auc = roc_auc_score(y_true, y_pred_prob[:, 1])
        print(f"\n{branch_name} AUC-ROC: {auc:.4f}")
    else:
        auc = roc_auc_score(y_true, y_pred_prob, multi_class='ovr')
        print(f"\n{branch_name} AUC-ROC (OVR): {auc:.4f}")

    return {
        "y_true": y_true,
        "y_pred": y_pred,
        "y_pred_prob": y_pred_prob,
        "confusion_matrix": cm,
        "auc": auc
    }

# --- Ensure test labels are one-hot for both branches ---
def get_flat_labels(y):
    y = np.asarray(y)
    if y.ndim > 1 and y.shape[-1] > 1:
        return np.argmax(y, axis=-1)
    return y.reshape(-1).astype(int)

from tensorflow.keras.utils import to_categorical
y_test_ecg_oh = to_categorical(get_flat_labels(y_test_ecg), num_classes=2)
y_test_pcg_oh = to_categorical(get_flat_labels(y_test_pcg), num_classes=2)

# --- Evaluate each branch independently ---
ecg_results = evaluate_branch(ecg_model, x_test_ecg, y_test_ecg_oh, branch_name="ECG")
pcg_results = evaluate_branch(pcg_model, x_test_pcg, y_test_pcg_oh, branch_name="PCG")

print(f"\n{'='*50}")
print("Summary")
print(f"{'='*50}")
print(f"ECG Branch AUC-ROC: {ecg_results['auc']:.4f}")
print(f"PCG Branch AUC-ROC: {pcg_results['auc']:.4f}")
```

```python
# =====================================================================
# DECISION-LEVEL MULTIMODAL FUSION (Proof-of-Concept — Index-Paired)
# =====================================================================
# LIMITATION (must be disclosed in your methodology/limitations section):
# x_test_ecg and x_test_pcg come from different patient cohorts
# (PTB-XL vs. PhysioNet/CirCor). No patient-level correspondence exists.
# Fusion below combines predictions at matching array indices as a
# proof-of-concept for the fusion architecture/strategy, NOT as a
# clinically valid multimodal diagnosis. Report unimodal AUCs (ecg_results,
# pcg_results) as primary results; report fused AUC as a secondary,
# clearly-caveated proof-of-concept number.
# =====================================================================

import numpy as np
from sklearn.metrics import roc_auc_score, classification_report, confusion_matrix

# --- Truncate to equal length for index-pairing (same as training-time approach) ---
min_test = min(len(x_test_ecg), len(x_test_pcg))
x_test_ecg_f = x_test_ecg[:min_test]
x_test_pcg_f = x_test_pcg[:min_test]
y_test_f = y_test_ecg[:min_test]   # using ECG labels as the fused target, consistent with earlier convention

y_test_f_flat = get_flat_labels(y_test_f)

# --- Get probability outputs from each independently-trained branch ---
ecg_probs_f = ecg_model.predict(x_test_ecg_f, verbose=0)   # shape (N, 2)
pcg_probs_f = pcg_model.predict(x_test_pcg_f, verbose=0)   # shape (N, 2)

ecg_pos_probs = ecg_probs_f[:, 1]
pcg_pos_probs = pcg_probs_f[:, 1]

# ============================================================
# FUSION STRATEGY 1: Simple averaging (equal weight)
# ============================================================
fused_probs_avg = 0.5 * ecg_pos_probs + 0.5 * pcg_pos_probs
fused_auc_avg = roc_auc_score(y_test_f_flat, fused_probs_avg)
print(f"[Fusion - Simple Average] AUROC: {fused_auc_avg:.4f}")

# ============================================================
# FUSION STRATEGY 2: Weighted average, weighted by each branch's
# own validation-set AUC (branches that perform better individually
# get more say in the fused decision)
# ============================================================
w_ecg = ecg_results['auc'] / (ecg_results['auc'] + pcg_results['auc'])
w_pcg = pcg_results['auc'] / (ecg_results['auc'] + pcg_results['auc'])

fused_probs_weighted = w_ecg * ecg_pos_probs + w_pcg * pcg_pos_probs
fused_auc_weighted = roc_auc_score(y_test_f_flat, fused_probs_weighted)
print(f"[Fusion - AUC-Weighted, w_ecg={w_ecg:.3f}, w_pcg={w_pcg:.3f}] AUROC: {fused_auc_weighted:.4f}")

# ============================================================
# FUSION STRATEGY 3: Max-confidence fusion (take whichever branch
# is more confident for each sample — a form of learned gating proxy)
# ============================================================
ecg_confidence = np.abs(ecg_pos_probs - 0.5)
pcg_confidence = np.abs(pcg_pos_probs - 0.5)
use_ecg = ecg_confidence >= pcg_confidence
fused_probs_maxconf = np.where(use_ecg, ecg_pos_probs, pcg_pos_probs)
fused_auc_maxconf = roc_auc_score(y_test_f_flat, fused_probs_maxconf)
print(f"[Fusion - Max Confidence] AUROC: {fused_auc_maxconf:.4f}")

# ============================================================
# Pick best strategy and report full metrics
# ============================================================
strategies = {
    "Simple Average": (fused_probs_avg, fused_auc_avg),
    "AUC-Weighted": (fused_probs_weighted, fused_auc_weighted),
    "Max Confidence": (fused_probs_maxconf, fused_auc_maxconf),
}
best_strategy_name = max(strategies, key=lambda k: strategies[k][1])
best_fused_probs, best_fused_auc = strategies[best_strategy_name]

print(f"\nBest fusion strategy: {best_strategy_name} (AUROC: {best_fused_auc:.4f})")

y_pred_fused = (best_fused_probs >= 0.5).astype(int)

print("\nFused Model Classification Report:")
print(classification_report(y_test_f_flat, y_pred_fused, target_names=["Normal", "Abnormal"]))
print("Fused Model Confusion Matrix:")
print(confusion_matrix(y_test_f_flat, y_pred_fused))

# ============================================================
# Final Summary Table
# ============================================================
print(f"\n{'='*50}\nFinal AUROC Summary\n{'='*50}")
print(f"ECG-only AUROC:              {ecg_results['auc']:.4f}")
print(f"PCG-only AUROC:              {pcg_results['auc']:.4f}")
print(f"Fused ({best_strategy_name}) AUROC: {best_fused_auc:.4f}  [proof-of-concept, index-paired]")
```

```python
# --- Independent Mamba+Transformer Branches (ECG, PCG) + Federated Learning + Decision Fusion ---
#
# Verified/updated against:
#   [1] Kavitha Bharathi S. et al., "A Federated Learning based Approach for
#       Heart Disease Prediction," ICCMC 2022 — Algorithm 1 (weighted FedAvg
#       aggregation formula).
#   [2] Pereira et al., "Healthcare Diagnostics Service Using Federated
#       Learning," ICONAT 2023 — explicit server-side validation split (1000+
#       images) kept distinct from the client training data AND the final
#       test set (Section IV).
#
# Fixes applied in this version:
#   (a) non_iid_partition rewritten as a shard-based scheme so every
#       requested client receives data regardless of how num_clients
#       compares to the number of classes. The previous version silently
#       left a client empty every round whenever num_clients > num_classes
#       with classes_per_client=1 (confirmed: with binary labels and
#       num_clients=3, client 3 got 0 samples in every round).
#   (b) Round-wise AUC checkpointing/early stopping now uses a validation
#       split carved from TRAIN data only. x_test/y_test are held out
#       completely until the very end, evaluated exactly once on the
#       restored best-checkpoint weights. This removes the test-set
#       selection-bias issue from the earlier version and matches [2]'s
#       explicit train/val/test separation.
#   (c) num_classes threaded as an explicit parameter instead of hardcoded
#       2, in case you extend to multi-class labels later.

import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, Model
import matplotlib.pyplot as plt
from sklearn.metrics import roc_auc_score
from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight
from tensorflow.keras.utils import to_categorical

# ============================================================
# 1. MAMBA (SELECTIVE SSM) BLOCK — tf.scan based (unchanged, verified correct)
# ============================================================
class MambaBlock(layers.Layer):
    def __init__(self, d_model, d_state=16, expand=2, **kwargs):
        super().__init__(**kwargs)
        self.d_model = d_model
        self.d_state = d_state
        self.d_inner = expand * d_model

        self.in_proj = layers.Dense(self.d_inner * 2, use_bias=False)
        self.conv1d = layers.Conv1D(
            filters=self.d_inner, kernel_size=4, padding='causal',
            groups=self.d_inner, use_bias=True, activation='swish'
        )
        self.x_proj = layers.Dense(self.d_state * 2 + self.d_inner, use_bias=False)
        self.dt_proj = layers.Dense(self.d_inner, use_bias=True)
        self.out_proj = layers.Dense(d_model, use_bias=False)

        self.A_log = self.add_weight(
            shape=(self.d_inner, self.d_state), initializer='ones',
            trainable=True, name='A_log'
        )
        self.D = self.add_weight(
            shape=(self.d_inner,), initializer='ones', trainable=True, name='D'
        )

    def call(self, x):
        batch = tf.shape(x)[0]

        x_and_res = self.in_proj(x)
        x_in, res = tf.split(x_and_res, 2, axis=-1)
        x_in = self.conv1d(x_in)

        x_dbc = self.x_proj(x_in)
        delta_raw, B, C = tf.split(
            x_dbc, [self.d_inner, self.d_state, self.d_state], axis=-1
        )
        delta = tf.nn.softplus(self.dt_proj(delta_raw))

        A = -tf.exp(self.A_log)

        deltaA = tf.exp(tf.einsum('bld,dn->bldn', delta, A))
        deltaB_x = tf.einsum('bld,bln,bld->bldn', delta, B, x_in)

        deltaA_t = tf.transpose(deltaA, [1, 0, 2, 3])
        deltaB_x_t = tf.transpose(deltaB_x, [1, 0, 2, 3])

        def scan_fn(h_prev, elems):
            dA_t, dBx_t = elems
            return dA_t * h_prev + dBx_t

        h0 = tf.zeros((batch, self.d_inner, self.d_state))
        h_states = tf.scan(scan_fn, (deltaA_t, deltaB_x_t), initializer=h0)
        h_states = tf.transpose(h_states, [1, 0, 2, 3])

        y = tf.einsum('bldn,bln->bld', h_states, C) + self.D * x_in
        y = y * tf.nn.swish(res)

        return self.out_proj(y)


# ============================================================
# 2. TRANSFORMER ENCODER BLOCK (unchanged)
# ============================================================
def transformer_encoder_block(x, d_model, num_heads=4, ff_dim=128, dropout=0.1):
    attn_out = layers.MultiHeadAttention(num_heads=num_heads, key_dim=d_model // num_heads)(x, x)
    attn_out = layers.Dropout(dropout)(attn_out)
    x = layers.LayerNormalization(epsilon=1e-6)(x + attn_out)

    ff = layers.Dense(ff_dim, activation='relu')(x)
    ff = layers.Dense(d_model)(ff)
    ff = layers.Dropout(dropout)(ff)
    return layers.LayerNormalization(epsilon=1e-6)(x + ff)


# ============================================================
# 3. UNIMODAL HYBRID MAMBA+TRANSFORMER MODEL BUILDERS
#    (each is a STANDALONE model: single input, single output —
#     no cross-modality concatenation, no assumption of pairing)
# ============================================================
def build_unimodal_model(input_shape, d_model=64, num_classes=2, lr=1e-3, name_prefix="branch"):
    inp = layers.Input(shape=input_shape, name=f"{name_prefix}_input")

    if len(input_shape) == 3:
        h, w, c = input_shape
        x = layers.Reshape((h, w * c))(inp)
    else:
        x = inp

    x = layers.Dense(d_model)(x)
    x = MambaBlock(d_model=d_model, d_state=16)(x)
    x = transformer_encoder_block(x, d_model)
    x = layers.GlobalAveragePooling1D()(x)
    x = layers.Dense(64, activation='relu')(x)
    x = layers.Dropout(0.3)(x)
    output = layers.Dense(num_classes, activation='softmax')(x)

    model = Model(inputs=inp, outputs=output)
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=lr),
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )
    return model

# NOTE: confirm these default shapes match your actual preprocessing —
# PTB-XL is commonly windowed as either (1000, 12) at 100Hz or (5000, 12)
# at 500Hz for 10s records; (500, 12) below assumes a shorter window/crop.
# Update if your pipeline uses a different length.
def build_ecg_model(input_shape=(500, 12), d_model=64, num_classes=2):
    return build_unimodal_model(input_shape, d_model=d_model, num_classes=num_classes, name_prefix="ecg")

def build_pcg_model(input_shape=(128, 128, 1), d_model=64, num_classes=2):
    return build_unimodal_model(input_shape, d_model=d_model, num_classes=num_classes, name_prefix="pcg")


# ============================================================
# 4. GENERIC HELPERS
# ============================================================
def get_flat_labels(y):
    y = np.asarray(y)
    if y.ndim > 1 and y.shape[-1] > 1:
        return np.argmax(y, axis=-1)
    return y.reshape(-1).astype(int)


def non_iid_partition(x, y, num_clients=3, seed=None):
    """
    Shard-based non-IID partitioning (McMahan-style), generalized to work
    for ANY num_clients relative to the number of classes.

    Each class is split into ceil(num_clients / num_classes) shards; all
    shards across all classes are pooled and dealt round-robin to clients,
    so every client is guaranteed at least one shard. For binary labels
    this still produces the extreme non-IID setting (most clients see
    predominantly one class), which is the scenario Pereira et al.
    flag as important future-work realism (Section V: "heterogeneous,
    non-IID data").

    Fixes a bug in the previous version: splitting `unique_labels` directly
    across num_clients meant any client index beyond num_classes got zero
    labels assigned, and therefore zero data, every round (confirmed for
    num_clients=3 with binary labels).
    """
    rng = np.random.default_rng(seed)
    y = np.asarray(y)
    unique_labels = np.unique(y)
    num_classes = len(unique_labels)
    shards_per_class = int(np.ceil(num_clients / num_classes))

    shards = []
    for label in unique_labels:
        idxs = np.where(y == label)[0]
        rng.shuffle(idxs)
        for chunk in np.array_split(idxs, shards_per_class):
            if len(chunk) > 0:
                shards.append((x[chunk], y[chunk]))

    rng.shuffle(shards)
    client_shards = [[] for _ in range(num_clients)]
    for i, shard in enumerate(shards):
        client_shards[i % num_clients].append(shard)

    partitions = []
    for shards_for_client in client_shards:
        if not shards_for_client:
            partitions.append((np.array([]), np.array([])))
            continue
        xs = np.concatenate([s[0] for s in shards_for_client], axis=0)
        ys = np.concatenate([s[1] for s in shards_for_client], axis=0)
        partitions.append((xs, ys))
    return partitions


def weighted_average_weights(weight_list, sample_counts):
    """Matches Algorithm 1's aggregation formula: w_{t+1} = sum_k (n_k/n) * w_k"""
    total = sum(sample_counts)
    avg_weights = []
    for layer_weights in zip(*weight_list):
        stacked = np.array([w * (n / total) for w, n in zip(layer_weights, sample_counts)])
        avg_weights.append(np.sum(stacked, axis=0))
    return avg_weights


def get_round_lr(base_lr, rnd, decay=0.9):
    return base_lr * (decay ** rnd)


def make_train_val_split(x, y, val_fraction=0.15, seed=42):
    """
    Carve a server-side validation split out of TRAIN data only.

    Mirrors Pereira et al. (ICONAT 2023), where round-wise metrics are
    computed on a validation set held at the server (1000+ images),
    distinct from both client training data and the final test set.
    x_test/y_test are never touched here, or anywhere before the final
    evaluation in run_federated_branch.
    """
    y_flat = get_flat_labels(y)
    x_tr, x_val, y_tr, y_val = train_test_split(
        x, y_flat, test_size=val_fraction, stratify=y_flat, random_state=seed
    )
    return x_tr, y_tr, x_val, y_val


# ============================================================
# 5. GENERIC FEDERATED TRAINING LOOP FOR ONE MODALITY
# ============================================================
def run_federated_branch(build_model_fn, x_train, y_train, x_test, y_test,
                          modality_name, num_clients=3, num_classes=2,
                          val_fraction=0.15,
                          num_rounds=15, local_epochs=3, batch_size=16,
                          base_lr=1e-3, patience=5, min_delta=1e-4, seed=42):

    # --- server-side validation split, carved out of TRAIN only ---
    x_tr, y_tr, x_val, y_val = make_train_val_split(x_train, y_train, val_fraction, seed)

    client_partitions = non_iid_partition(x_tr, y_tr, num_clients, seed=seed)
    for i in range(len(client_partitions)):
        xc, yc = client_partitions[i]
        if len(yc) > 0:
            client_partitions[i] = (xc, to_categorical(yc, num_classes=num_classes))

    y_val_oh = to_categorical(get_flat_labels(y_val), num_classes=num_classes)
    y_val_flat = get_flat_labels(y_val_oh)

    global_model = build_model_fn()
    global_weights = global_model.get_weights()

    best_auc, best_weights, rounds_no_improve = -np.inf, global_weights, 0
    history = {"acc": [], "loss": [], "auc": []}

    for rnd in range(num_rounds):
        print(f"[{modality_name}] Round {rnd+1}/{num_rounds}")
        client_weights, sample_counts = [], []
        current_lr = get_round_lr(base_lr, rnd)

        for cid in range(num_clients):
            xc, yc = client_partitions[cid]
            if len(xc) == 0:
                continue
            yc_flat = get_flat_labels(yc)
            classes = np.unique(yc_flat)
            class_weight = None
            if len(classes) > 1:
                cw = compute_class_weight('balanced', classes=classes, y=yc_flat)
                class_weight = {int(c): w for c, w in zip(classes, cw)}

            model = build_model_fn()
            model.set_weights(global_weights)
            model.optimizer.learning_rate.assign(current_lr)
            model.fit(xc, yc, epochs=local_epochs, batch_size=batch_size,
                      class_weight=class_weight, verbose=0)
            client_weights.append(model.get_weights())
            sample_counts.append(len(yc))

        if not client_weights:
            print("  No clients participated.")
            continue

        global_weights = weighted_average_weights(client_weights, sample_counts)
        global_model.set_weights(global_weights)

        # --- round-wise metrics on the VALIDATION split only; test stays untouched ---
        loss, acc = global_model.evaluate(x_val, y_val_oh, verbose=0)
        val_probs = global_model.predict(x_val, verbose=0)
        auc = (roc_auc_score(y_val_flat, val_probs[:, 1]) if val_probs.shape[-1] == 2
               else roc_auc_score(y_val_flat, val_probs.ravel()))

        history["acc"].append(acc); history["loss"].append(loss); history["auc"].append(auc)
        print(f"  LR:{current_lr:.6f} ValLoss:{loss:.4f} ValAcc:{acc:.4f} ValAUC:{auc:.4f}")

        if auc - best_auc > min_delta:
            best_auc, best_weights, rounds_no_improve = auc, global_weights, 0
        else:
            rounds_no_improve += 1
        if rounds_no_improve >= patience:
            print(f"  Early stop at round {rnd+1}")
            break

    # --- restore best checkpoint, then touch the TEST set exactly once ---
    global_model.set_weights(best_weights)
    y_test_oh = to_categorical(get_flat_labels(y_test), num_classes=num_classes)
    y_test_flat = get_flat_labels(y_test_oh)
    test_probs = global_model.predict(x_test, verbose=0)
    test_auc = (roc_auc_score(y_test_flat, test_probs[:, 1]) if test_probs.shape[-1] == 2
                else roc_auc_score(y_test_flat, test_probs.ravel()))

    print(f"[{modality_name}] Best VAL ROC-AUC during training: {best_auc:.4f}")
    print(f"[{modality_name}] Final held-out TEST ROC-AUC (single evaluation): {test_auc:.4f}")

    return global_model, history, test_auc, y_test_flat, test_probs


# ============================================================
# 6. RUN BOTH BRANCHES INDEPENDENTLY
#    (x_train_ecg/y_train_ecg and x_train_pcg/y_train_pcg must come
#     from SEPARATE train_test_splits — do NOT truncate to min_samples)
# ============================================================
ecg_model, ecg_history, ecg_test_auc, ecg_y_flat, ecg_probs = run_federated_branch(
    build_ecg_model, x_train_ecg, y_train_ecg, x_test_ecg, y_test_ecg,
    modality_name="ECG", num_clients=3
)

pcg_model, pcg_history, pcg_test_auc, pcg_y_flat, pcg_probs = run_federated_branch(
    build_pcg_model, x_train_pcg, y_train_pcg, x_test_pcg, y_test_pcg,
    modality_name="PCG", num_clients=3
)

print(f"\nFinal ECG-branch Test ROC-AUC: {ecg_test_auc:.4f}")
print(f"Final PCG-branch Test ROC-AUC: {pcg_test_auc:.4f}")

ecg_model.save('ECG_Mamba_Federated_Model.keras')
pcg_model.save('PCG_Mamba_Federated_Model.keras')

# ============================================================
# 7. DECISION-LEVEL FUSION
# ============================================================
# 7a. GENUINE fusion — only valid with real paired patients.
#     If you obtain a set of patients with BOTH an ECG and a PCG
#     recording, evaluate fusion only on that subset:
#
# ecg_probs_paired = ecg_model.predict(x_paired_ecg, verbose=0)[:, 1]
# pcg_probs_paired = pcg_model.predict(x_paired_pcg, verbose=0)[:, 1]
# fused_probs = 0.5 * ecg_probs_paired + 0.5 * pcg_probs_paired
# fused_auc = roc_auc_score(y_paired, fused_probs)
# print(f"Fused (paired-patient) ROC-AUC: {fused_auc:.4f}")


def run_synthetic_pairing_fusion_demo(ecg_model, pcg_model,
                                       x_test_ecg, y_test_ecg,
                                       x_test_pcg, y_test_pcg,
                                       seed=42):
    """
    7b. PROOF-OF-CONCEPT FUSION DEMO — NOT A VALID PERFORMANCE METRIC.

    PTB-XL (ECG) and CirCor (PCG) share no real patients, so there is no
    ground truth for "the same person's" ECG+PCG pair. This function
    randomly pairs unrelated ECG and PCG test samples purely to exercise
    the fusion mechanism end-to-end (predict -> average -> AUROC), so you
    can confirm the fusion code path works and see a representative
    number shape.

    The reported AUROC here does NOT measure real multimodal diagnostic
    performance and must never be placed next to ecg_test_auc /
    pcg_test_auc as if it were a comparable, genuine result. If you use
    this number anywhere in the paper, it must be labeled explicitly as
    a synthetic-pairing architecture demonstration, with a note that no
    patient in the "fused" set actually has both signals.

    Ground truth for scoring is taken arbitrarily from the ECG side of
    each random pair (an explicit modeling choice made only so the demo
    has a well-defined AUROC to compute — it does not imply the paired
    PCG sample belongs to that same patient).
    """
    rng = np.random.default_rng(seed)
    n_pairs = min(len(y_test_ecg), len(y_test_pcg))

    ecg_idx = rng.permutation(len(y_test_ecg))[:n_pairs]
    pcg_idx = rng.permutation(len(y_test_pcg))[:n_pairs]

    ecg_probs = ecg_model.predict(x_test_ecg[ecg_idx], verbose=0)
    pcg_probs = pcg_model.predict(x_test_pcg[pcg_idx], verbose=0)

    ecg_pos = ecg_probs[:, 1] if ecg_probs.shape[-1] == 2 else ecg_probs.ravel()
    pcg_pos = pcg_probs[:, 1] if pcg_probs.shape[-1] == 2 else pcg_probs.ravel()
    fused_probs = 0.5 * ecg_pos + 0.5 * pcg_pos

    y_ground_truth = get_flat_labels(y_test_ecg)[ecg_idx]
    fused_auc = roc_auc_score(y_ground_truth, fused_probs)

    print("\n" + "=" * 72)
    print("PROOF-OF-CONCEPT FUSION DEMO — SYNTHETIC RANDOM PAIRING")
    print("This is NOT a real multimodal performance metric. Do not cite")
    print("this number as comparable to ecg_test_auc / pcg_test_auc without")
    print("the synthetic-pairing disclosure described in this function's")
    print("docstring.")
    print(f"  Synthetic pairs evaluated: {n_pairs}")
    print(f"  Synthetic fused ROC-AUC:   {fused_auc:.4f}")
    print("=" * 72 + "\n")

    return fused_auc


# Set to True only if you want to see the fusion demo number.
# Leave False for the paper's primary results (ECG/PCG reported separately).
RUN_SYNTHETIC_FUSION_DEMO = False

if RUN_SYNTHETIC_FUSION_DEMO:
    synthetic_fused_auc = run_synthetic_pairing_fusion_demo(
        ecg_model, pcg_model, x_test_ecg, y_test_ecg, x_test_pcg, y_test_pcg
    )

plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(ecg_history["auc"], marker='o', label='ECG (Mamba+Transformer)')
plt.plot(pcg_history["auc"], marker='s', label='PCG (Mamba+Transformer)')
plt.title('Validation ROC-AUC per Round (Independent Branches)')
plt.xlabel('Round'); plt.ylabel('ROC-AUC'); plt.legend(); plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(ecg_history["loss"], marker='o', label='ECG')
plt.plot(pcg_history["loss"], marker='s', label='PCG')
plt.title('Validation Loss per Round (Independent Branches)')
plt.xlabel('Round'); plt.ylabel('Loss'); plt.legend(); plt.grid(True)
plt.tight_layout()
plt.show()
```
